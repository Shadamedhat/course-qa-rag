```python
def clean_text(text):
    text = text.replace("ﬁ", "fi")
    text = text.replace("ﬂ", "fl")
    text = text.replace("■", "")
    text = " ".join(text.split())
    return text
```

```python
def split_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]
        chunks.append(chunk)
        start = end - overlap
    return chunks
```

```python
test_text = "a" * 1200
test_chunks = split_text(test_text)

print("Number of chunks:", len(test_chunks))
for c in test_chunks:
    print(len(c))
```

```python
chunks = []

for doc in documents:
    if len(doc["text"]) <= 500:
        chunks.append(doc)
    else:
        pieces = split_text(doc["text"])
        for piece in pieces:
            new_doc = doc.copy()
            new_doc["text"] = piece
            chunks.append(new_doc)

print("Number of documents before:", len(documents))
print("Number of chunks after:", len(chunks))
print(chunks[0])
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

test_vector = model.encode("What is a matrix?")

print("Vector length:", len(test_vector))
print(test_vector[:5])
```

```python
texts = []
for chunk in chunks:
    texts.append(chunk["text"])

embeddings = model.encode(texts, show_progress_bar=True)

print("Number of texts:", len(texts))
print("Embeddings shape:", embeddings.shape)
```

```python
import chromadb

db_path = "/content/drive/MyDrive/rag_project/chroma_db"

client = chromadb.PersistentClient(path=db_path)
collection = client.get_or_create_collection(name="course_materials")

print("Collection created:", collection.name)
```

```python
ids = []
metadatas = []

for i in range(len(chunks)):
    chunk = chunks[i]

    page = chunk["page"]
    if page is None:
        page = 0

    row = chunk.get("row", 0)

    ids.append("chunk_" + str(i))
    metadatas.append({
        "course": chunk["course"],
        "source": chunk["source"],
        "page": page,
        "row": row
    })

collection.upsert(
    ids=ids,
    documents=texts,
    embeddings=embeddings.tolist(),
    metadatas=metadatas
)

print("Number of items in the database:", collection.count())
```

```python
question = "What is the column space of a matrix?"

question_embedding = model.encode(question)

results = collection.query(
    query_embeddings=[question_embedding.tolist()],
    n_results=3
)

for i in range(3):
    print("Result", i + 1)
    print("Source:", results["metadatas"][0][i]["source"])
    print("Page:", results["metadatas"][0][i]["page"])
    print("Text:", results["documents"][0][i][:200])
    print("-----")
```

```python
def search(question, n_results=3):
    question_embedding = model.encode(question)
    results = collection.query(
        query_embeddings=[question_embedding.tolist()],
        n_results=n_results
    )
    return results
```

```python
questions = [
    "What is the difference between a list and a tuple?",
    "What is the difference between precision and recall?"
]

for q in questions:
    results = search(q, n_results=2)
    print("Question:", q)
    for i in range(2):
        print("   Source:", results["metadatas"][0][i]["source"])
        print("   Text:", results["documents"][0][i][:100])
    print("-----")
```

```python
def build_context(results):
    context = ""
    for i in range(len(results["documents"][0])):
        text = results["documents"][0][i]
        source = results["metadatas"][0][i]["source"]
        context = context + "[Source: " + source + "] " + text + "\n\n"
    return context


def build_prompt(question, context):
    prompt = "Answer the question using ONLY the context below. "
    prompt = prompt + "If the answer is not in the context, say: I could not find the answer in the course materials. "
    prompt = prompt + "Write the answer in plain text only, without markdown or LaTeX.\n\n"
    prompt = prompt + "Context:\n" + context
    prompt = prompt + "Question: " + question + "\nAnswer:"
    return prompt


question = "What is the column space of a matrix?"
results = search(question)
context = build_context(results)
prompt = build_prompt(question, context)

print(prompt)
```

```python
import time
from groq import Groq
from google.colab import userdata

groq_key = userdata.get("GROQ_API_KEY")
groq_client = Groq(api_key=groq_key, timeout=30)

groq_model = "openai/gpt-oss-120b"


def ask_llm(prompt):
    for attempt in range(3):
        try:
            response = groq_client.chat.completions.create(
                model=groq_model,
                messages=[{"role": "user", "content": prompt}]
            )
            return response.choices[0].message.content
        except Exception as e:
            print("Attempt", attempt + 1, "failed:", e)
            time.sleep(5)
    return "The model is busy right now. Please try again later."


print(ask_llm("Say hello in one short sentence."))
```

```python
def format_sources(results):
    sources = []
    for i in range(len(results["metadatas"][0])):
        meta = results["metadatas"][0][i]

        line = meta["source"]
        if meta["page"] != 0:
            line = line + " (page " + str(meta["page"]) + ")"
        if meta["row"] != 0:
            line = line + " (row " + str(meta["row"]) + ")"

        if line not in sources:
            sources.append(line)

    return sources
```

```python
def ask_question(question):
    results = search(question, n_results=3)
    context = build_context(results)
    prompt = build_prompt(question, context)

    answer = ask_llm(prompt)
    sources = format_sources(results)

    if "could not find the answer" in answer.lower():
        sources = []

    return answer, sources
```

```python
test_questions = [
    "What is the column space of a matrix?",
    "What is the difference between a list and a tuple?",
    "What is the difference between precision and recall?",
    "Who won the World Cup?"
]

for q in test_questions:
    answer, sources = ask_question(q)
    print("Question:", q)
    print("Answer:", answer)
    print("Sources:")
    for s in sources:
        print("  -", s)
    print("==========")
```

```python
import gradio as gr

def answer_for_ui(question):
    answer, sources = ask_question(question)

    if len(sources) == 0:
        sources_text = "No sources"
    else:
        sources_text = "\n".join(sources)

    return answer, sources_text


app = gr.Interface(
    fn=answer_for_ui,
    inputs=gr.Textbox(label="Ask a question about the courses", lines=2),
    outputs=[
        gr.Textbox(label="Answer", lines=8),
        gr.Textbox(label="Sources", lines=4)
    ],
    title="Course Q&A RAG System"
)

app.launch(debug=True)
```
